```python
# pip install matplotlib
import matplotlib.pyplot as plt
from matplotlib.widgets import RectangleSelector

def onselect(eclick, erelease):
    """Callback for rectangle selection."""
    x1, y1 = eclick.xdata, eclick.ydata   # starting point
    x2, y2 = erelease.xdata, erelease.ydata  # ending point
    print(f"Rectangle: ({x1:.2f}, {y1:.2f}) --> ({x2:.2f}, {y2:.2f})")
    print(f"Width: {abs(x2 - x1):.2f}, Height: {abs(y2 - y1):.2f}")

# Create a sample image
fig, ax = plt.subplots()
ax.imshow([[0,1],[1,0]], cmap='gray', extent=[0, 100, 0, 100])
ax.set_title("Drag to draw a rectangle")

# Enable RectangleSelector
selector = RectangleSelector(
    ax, onselect,
    useblit=True,
    button=[1],            # left mouse button
    minspanx=5, minspany=5,
    spancoords='data',
    interactive=True
)

plt.show()
```

```python
# pip install matplotlib
import matplotlib
matplotlib.use("TkAgg")  # <— ensures an interactive GUI window
import matplotlib.pyplot as plt
from matplotlib.widgets import RectangleSelector

def onselect(eclick, erelease):
    x1, y1 = eclick.xdata, eclick.ydata
    x2, y2 = erelease.xdata, erelease.ydata
    print(f"Rect: ({x1:.1f}, {y1:.1f}) → ({x2:.1f}, {y2:.1f})  "
          f"W={abs(x2-x1):.1f}, H={abs(y2-y1):.1f}")

fig, ax = plt.subplots()
ax.imshow([[0,1],[1,0]], cmap="gray", extent=[0,100,0,100])
ax.set_title("Drag to draw a rectangle (Esc to cancel)")

selector = RectangleSelector(
    ax, onselect,
    useblit=True,
    button=[1],          # left click
    minspanx=5, minspany=5,
    spancoords="data",
    interactive=True
)
selector.set_active(True)  # make sure it's enabled

plt.show()  # GUI window; drag to draw
```
